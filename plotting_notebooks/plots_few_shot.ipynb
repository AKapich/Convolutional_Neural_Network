```python
import json
import matplotlib.pyplot as plt
```

```python
path_r = ".\\few_shot\\results\\resnet\\resnet_log.json"
path_v = ".\\few_shot\\results\\vgg16\\vgg_log.json"
path_c = ".\\few_shot\\results\\custom\\custom_log.json"


with open(path_r, "r") as f:
    data_r = json.load(f)

with open(path_v, "r") as f:
    data_v = json.load(f)

with open(path_c, "r") as f:
    data_c = json.load(f)
```

```python
epochs = [item["epoch"] for item in data_r]

val_accuracy_r = [item["val_accuracy"] for item in data_r]
val_accuracy_v = [item["val_accuracy"] for item in data_v]
val_accuracy_c = [item["val_accuracy"] for item in data_c]

plt.figure(figsize=(6, 3.5))
plt.plot(epochs, val_accuracy_r, marker="o", linestyle="-", color="b", label="ResNet")
plt.plot(epochs, val_accuracy_v, marker="s", linestyle="-", color="g", label="VGG16")
plt.plot(epochs, val_accuracy_c, marker="^", linestyle="-", color="r", label="Custom")

plt.title("Comparison of val. accuracy between models")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy (%)")
plt.grid(True, linestyle="--", alpha=0.7)
plt.xticks(epochs)
plt.legend()
plt.tight_layout()
plt.show()
```
